```python
import pandas as pd
import numpy as np
```

```python
#sales_data = pd.read_csv(r"C:\Users\wwwsi\OneDrive\Documents\Project Brothers BBQ\Dataset _\Total_Monthly_Sales.csv")
Transction_data = pd.read_csv(r"C:\Users\wwwsi\OneDrive\Documents\Project Brothers BBQ\Dataset _\raw\sales_transactions.csv")
```

```python
Transction_data.head()
```

```python
Transction_data.info()
```

```python
Transction_data['date'] = pd.to_datetime(Transction_data["date"],errors = 'coerce')
Transction_data['time_of_purchase'] = pd.to_datetime(Transction_data['time_of_purchase'],errors = 'coerce')
Transction_data['age'] = Transction_data['age'].astype("Int64")
Transction_data['Item_price_total'] = Transction_data['item_price_total'].astype("float64")
Transction_data['discount_amount'] = Transction_data['discount_amount'].astype("float64")
Transction_data['final_amount'] = Transction_data['final_amount'].astype("float64")
```

```python
Transction_data.info()
```

```python
missing_values = ["N/A", "NA", "na", "NaN", "NULL", "null", "Missing", ""]
Transction_data.isin(missing_values).sum()
```

```python
# Search for missing values
print(Transction_data.shape)
Transction_data.isnull().sum()
```

```python
# Filling Missing Values
# Assuming IF Card or UPI like digital payment have the Transcation history and Details
# Hard-Cash will be the one most likely have the possibility forget to entry
Transction_data["payment_method"] = Transction_data["payment_method"].fillna('Cash') 
```

```python
Transction_data['rating'] = Transction_data['rating'].fillna('Not rated')
#Transction_data['age'] = Transction_data['age'].fillna('Unknown')
```

```python
Transction_data['customer_name'] = (Transction_data.groupby('phone_number')['customer_name'].transform(lambda x:x.ffill().bfill()))
Transction_data['customer_name'] = Transction_data['customer_name'].fillna("Unknown")
```

```python
Transction_data['age_num'] = Transction_data['age'].replace('Unknown',np.nan).astype("Int64")
```

```python
cols = list(Transction_data.columns)
cols.remove("age_num")
cols.insert(cols.index("age")+1,"age_num")
cols
```

```python
Transction_data = Transction_data[cols]
Transction_data.head(10)
```

```python
#removing redunency columns
Transction_data.drop(['coupon_code','review'],axis = 1,inplace = True)
```

```python
# Reload the original raw CSV just to get a clean time_of_purchase column
raw = pd.read_csv(r"C:\Users\wwwsi\OneDrive\Documents\Project Brothers BBQ\Dataset _\raw\sales_transactions.csv")

# Parse time-only, no format guessing, no date attached
Transction_data['time_of_purchase'] = pd.to_datetime(
    raw['time_of_purchase'], format='%H:%M', errors='coerce'
).dt.time

# Now combine with date into one real timestamp
Transction_data['purchase_datetime'] = pd.to_datetime(
    Transction_data['date'].astype(str) + ' ' + Transction_data['time_of_purchase'].astype(str),
    errors='coerce'
)

print(Transction_data[['date', 'time_of_purchase', 'purchase_datetime']].head())
print(Transction_data['purchase_datetime'].isnull().sum(), "rows failed to parse")
```
