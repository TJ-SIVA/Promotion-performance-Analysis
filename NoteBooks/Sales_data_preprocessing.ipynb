```python
sales_data["date"] = pd.to_datetime(sales_data["date"],errors = 'coerce')
sales_data["total_sales_amount"]=sales_data["total_sales_amount"].astype("float64")
sales_data.info()
```

```python
#Spotting outliers
Numerical_column = sales_data.select_dtypes(include=['number']).columns
for x in Numerical_column:
    Q1 = sales_data[x].quantile(0.25)
    Q3 = sales_data[x].quantile(0.75)
    IQR = Q3 - Q1

    #Determing outlier
    lower_bound = Q1-1.5*IQR
    upper_bound = Q3+1.5*IQR

    outlier = sales_data[(sales_data[x] < lower_bound) | (sales_data[x] > upper_bound)]
    print(f'\n{ x } :{outlier[x].count()}')
```

```python
# Validating Data quality
# 1. Validating total stocks with splited stocks [ leg + wings + lolipop = total_stocks]
pd.DataFrame(np.isclose(sales_data['total_stock_kg'],
                        sales_data[['leg_stock_kg','wings_stock_kg','lollipop_stock_kg']].sum(axis=1),
                        atol =1e-8)).value_counts()
```

```python
# 2. Validating Planned_Customer <= Served_Customer
(sales_data['planned_customers'] <= sales_data['actual_customers_served'] ).all()
```

```python
# Creating new column for Total pieces count
sales_data['total_pieces'] = sales_data[['leg_pieces_sold','wings_pieces_sold','lollipop_pieces_sold',
                                         'leg_pieces_leftover','wings_pieces_leftover','lollipop_pieces_leftover']].sum(axis = 1)

# Createing new column for total sold pieces count
sales_data['total_sold_pieces'] = sales_data['total_pieces'] - (sales_data[['leg_pieces_leftover',
                                                                            'wings_pieces_leftover',
                                                                            'lollipop_pieces_leftover']].sum(axis=1))

# Creating new column for total number of pieces leftover

sales_data['total_sold_pieces'] = sales_data['total_pieces'] - (sales_data[['leg_pieces_sold',
                                                                            'wings_pieces_sold',
                                                                            'lollipop_pieces_sold']].sum(axis=1))

# Createing new column for total kg of stock leftover
sales_data['total_leftover_kg'] = sales_data[['leg_leftover_kg','wings_leftover_kg','lollipop_leftover_kg']].sum(axis=1)

# Creating new column for total kg of stock sold

sales_data['total_sold_kg'] = sales_data['total_stock_kg'] - sales_data['total_leftover_kg'] 

```

```python
sales_data = sales_data.rename(columns={'toatl_pieces':'total_pieces'})
```
